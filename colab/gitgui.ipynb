# gitgui in Google Colab

Startet den gitgui-Server in der Colab-Runtime und zeigt die GUI (Guided + Alle Befehle) in einem IFrame an.
git läuft damit in der Colab-Linux-Umgebung — Repos liegen unter `/content`.

**Wichtig:** Ersetze `REPO_URL` unten durch die URL dieses Projekts (oder lade die Dateien anderweitig nach `/content/gitgui` hoch).

```python
REPO_URL = "https://github.com/OWNER/gitgui.git"  # <-- anpassen
TOKEN = "colab-token"  # <-- eigenes Token setzen
!git clone --depth 1 {REPO_URL} /content/gitgui 2>/dev/null || cp -r /content/drive/MyDrive/gitgui /content/gitgui
!cd /content/gitgui && python3 -c "import sys; assert sys.version_info >= (3, 8)" && echo ok
```

```python
import subprocess, time
proc = subprocess.Popen(
    ["python3", "server.py", "--host", "127.0.0.1", "--port", "8765",
     "--token", TOKEN, "--state-dir", "/content/gitgui-state"],
    cwd="/content/gitgui", stdout=open("/tmp/gitgui-server.log", "w"), stderr=subprocess.STDOUT)
time.sleep(2)
print(open("/tmp/gitgui-server.log").read())
```

```python
from google.colab import output
output.serve_kernel_port_as_iframe(8765, path="/?token=" + TOKEN, height=900)
```

## Nutzung

- Im IFrame oben öffnet sich die GUI. Das Token hängt als `?token=…` an der URL (nur lokal in dieser Colab-Session erreichbar).
- Repos liegen unter `/content` (z. B. `/content/gitgui` selbst). Im „Guided“-Tab kann man über „+ repo“ ein Verzeichnis wählen.
- Die Android-App kann dieselbe Session fernsteuern: Backend `http` mit Server-URL und Token.
- Aufräumen: `proc.terminate()` in der Server-Zelle.
